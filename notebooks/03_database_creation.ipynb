```python
import pandas as pd
from sqlalchemy import create_engine

# Connect to database
engine = create_engine("sqlite:///../data/bluestock_mf.db")

# Load cleaned files
fund_master = pd.read_csv("../data/processed/01_fund_master_clean.csv")
nav_history = pd.read_csv("../data/processed/02_nav_history_clean.csv")

# Save to SQLite
fund_master.to_sql(
    "dim_fund",
    engine,
    if_exists="replace",
    index=False
)

nav_history.to_sql(
    "fact_nav",
    engine,
    if_exists="replace",
    index=False
)

print("Tables loaded successfully!")
```

```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("sqlite:///../bluestock_mf.db")

fund_count = pd.read_sql("SELECT COUNT(*) as count FROM dim_fund", engine)
nav_count = pd.read_sql("SELECT COUNT(*) as count FROM fact_nav", engine)

print("dim_fund rows:")
print(fund_count)

print("\nfact_nav rows:")
print(nav_count)
```
